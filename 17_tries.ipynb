<h3>1. Implement TRIE | INSERT | SEARCH | STARTSWITH</h3>
<a href="https://leetcode.com/problems/implement-trie-prefix-tree/">Problem Link</a>
<p> 
Data Structure Design:
A Trie can be represented using a tree-like structure where each node corresponds to a character in a string. Each node has:

A children dictionary where the keys are characters, and the values are child nodes corresponding to the next character in the word.
A boolean flag (often is_end_of_word) to indicate whether a node represents the end of a valid word.


Operations:
Insert: Traverse the Trie character by character, creating new nodes if necessary. Mark the last node as the end of the word.
Search: Traverse the Trie using the characters of the word. If we reach the end of the word and the last node is marked as is_end_of_word, return True; otherwise, return False.
StartsWith: Similar to search, but we don’t need to check if the node is the end of the word; just verify if the prefix exists.

<br><br>
Time complexity: O(n)<br>
Space Complexity: O(n * m)</p>

```python
class Trie:

    def __init__(self):
        # The Trie is initialized with an empty root node
        self.root = {}
        
    def insert(self, word: str) -> None:
        # Starting from the root, insert each character of the word
        node = self.root
        for char in word:
            if char not in node:
                node[char] = {}  # Create a new node if it doesn't exist
            node = node[char]  # Move to the next level
        node['#'] = True  # Mark the end of the word with a special character
        
    def search(self, word: str) -> bool:
        # Check if the word exists in the Trie
        node = self.root
        for char in word:
            if char not in node:
                return False  # Return False if the character is not found
            node = node[char]  # Move to the next level
        return '#' in node  # Return True if end of word is marked with '#'
        
    def startsWith(self, prefix: str) -> bool:
        # Check if any word starts with the given prefix
        node = self.root
        for char in prefix:
            if char not in node:
                return False  # Return False if the character is not found
            node = node[char]  # Move to the next level
        return True  # Return True if all characters in the prefix are found
```

<h3>1. Implement Trie - 2 (Prefix Tree)</h3>
<a href="https://www.naukri.com/code360/problems/implement-trie_1387095?utm_source=youtube&utm_medium=affiliate&utm_campaign=striver_tries_videos">Problem Link</a>
<p> 
Trie Node:
A node in the Trie will have:
A dictionary of children nodes (children).
An integer to store the count of times a word ends at this node (count).
An integer to store the number of words that pass through this node (prefix_count).

Trie Operations:
Insert: Traverse the trie for each character of the word, creating new nodes if needed, and increment the counts.
CountWordsEqualTo: Traverse the trie to find the word and return the count at the end node.
CountWordsStartingWith: Traverse the trie to find the node corresponding to the prefix and return the prefix_count at that node.
Erase: Decrease the count at each node for the word, and delete the node if it's no longer needed.
<br><br>
Time complexity: O(m)<br>
Space Complexity: O(k)</p>

```python
class Trie:
    def __init__(self):
        # Initialize an empty trie
        self.root = {}
    
    def insert(self, word: str):
        # Insert word into the trie
        node = self.root
        for char in word:
            if char not in node:
                node[char] = {}  # Create a new node if it doesn't exist
            node = node[char]
            node['prefix_count'] = node.get('prefix_count', 0) + 1  # Track the prefix count
        node['count'] = node.get('count', 0) + 1  # Track the count of this word
    
    def countWordsEqualTo(self, word: str) -> int:
        # Count how many times the word is in the trie
        node = self.root
        for char in word:
            if char not in node:
                return 0  # Word does not exist in the trie
            node = node[char]
        return node.get('count', 0)  # Return the count of the word
    
    def countWordsStartingWith(self, prefix: str) -> int:
        # Count how many words start with the given prefix
        node = self.root
        for char in prefix:
            if char not in node:
                return 0  # No words start with this prefix
            node = node[char]
        return node.get('prefix_count', 0)  # Return the prefix count
    
    def erase(self, word: str):
        # Erase one occurrence of the word from the trie
        node = self.root
        stack = []
        
        # First, we traverse the trie and record the nodes along the path of the word
        for char in word:
            if char not in node:
                return  # Word doesn't exist (this should never happen in this problem)
            stack.append((node, char))
            node = node[char]
        
        # Decrease the count of the word at the last node
        if node.get('count', 0) > 0:
            node['count'] -= 1
        
        # Now we need to propagate the changes and possibly clean up nodes
        for parent, char in reversed(stack):
            if parent[char]['prefix_count'] > 0:
                parent[char]['prefix_count'] -= 1
            if parent[char]['prefix_count'] == 0 and 'count' not in parent[char]:
                del parent[char]  # Delete the node if it is no longer needed
```

<h3>2. Longest String with All Prefixes</h3>
<a href="https://www.naukri.com/code360/problems/complete-string_2687860?utm_source=youtube&utm_medium=affiliate&utm_campaign=striver_tries_videos">Problem Link</a>
<p> 
TrieNode Class:

Each node in the Trie has:
children: A dictionary to store child nodes corresponding to each character.
is_end_of_word: A boolean flag to mark the end of a valid word.

Trie Class:

insert(word):
Adds a word to the Trie by iterating over its characters.
Creates new child nodes if the character is not already present.
Marks the last node as the end of the word.
is_complete_string(word):
Checks if a word is a complete string.
A word is a complete string if all its prefixes are present in the Trie and are marked as valid words (is_end_of_word).

completeString Function:

Creates a Trie and inserts all the words in the array.
Iterates over all the words to check if they are complete strings using is_complete_string.
Maintains the longest complete string or the lexicographically smallest string in case of ties.

<br><br>
Time complexity: O(sum_len)<br>
Space Complexity: O(sum_len)</p>

```python
def completeString(n, a):
    class TrieNode:
        def __init__(self):
            self.children = {}
            self.is_end_of_word = False

    class Trie:
        def __init__(self):
            self.root = TrieNode()

        def insert(self, word: str):
            node = self.root
            for char in word:
                if char not in node.children:
                    node.children[char] = TrieNode()
                node = node.children[char]
            node.is_end_of_word = True

        def is_complete_string(self, word: str) -> bool:
            node = self.root
            for char in word:
                if char not in node.children or not node.children[char].is_end_of_word:
                    return False
                node = node.children[char]
            return True

    # Initialize the Trie and insert all words
    trie = Trie()
    for word in a:
        trie.insert(word)

    # Find the longest complete string
    longest_complete_str = None
    for word in a:
        if trie.is_complete_string(word):
            if (longest_complete_str is None or 
                len(word) > len(longest_complete_str) or 
                (len(word) == len(longest_complete_str) and word < longest_complete_str)):
                longest_complete_str = word

    return longest_complete_str if longest_complete_str else "None"
```

<h3>3. Number of Distinct Substrings in a String</h3>
<a href="https://www.naukri.com/code360/problems/count-distinct-substrings_985292?utm_source=youtube&utm_medium=affiliate&utm_campaign=striver_tries_videos">Problem Link</a>
<p> 
Key Insight:

Every substring of S is a prefix of some suffix of S.
By inserting all suffixes of S into a Trie, we can automatically count the distinct substrings.

Algorithm:

For each suffix of S:
Insert it into the Trie character by character.
Count the number of new nodes created during this process (each new node corresponds to a new substring).
Add 1 to account for the empty substring.

Implementation:

Use a Trie structure with nodes storing child references for characters.
Insert suffixes and count new nodes during insertion.
<br><br>
Time complexity: O(n^2)<br>
Space Complexity: O(n^2)</p>

```python
class TrieNode:
    def __init__(self):
        self.children = {}

def countDistinctSubstrings(s):
    root = TrieNode()
    count = 0  # To count distinct substrings

    for i in range(len(s)):
        current = root
        for j in range(i, len(s)):
            char = s[j]
            if char not in current.children:
                current.children[char] = TrieNode()
                count += 1  # New node means a new substring
            current = current.children[char]

    return count + 1  # +1 for the empty substring
```

<h3>4. Bit PreRequisites for TRIE Problems</h3>
<a href="https://www.geeksforgeeks.org/problems/bits-basic-operations/1?utm_source=youtube&utm_medium=collab_striver_ytdescription&utm_campaign=bits-basic-operations">Problem Link</a>
<p> 
XOR Function:

The XOR operation is performed using the ^ operator in Python.
The result is the bitwise XOR of n and m.

check Function:

To check if the a-th bit (1-indexed) of b is set:
Use the bitwise AND (&) operation with a bitmask where only the 
(a−1)-th bit is set. This is constructed as (1<<(a−1)).
If the result is non-zero, the a-th bit is set; otherwise, it is not.

setBit Function:

To set the 
c-th bit (0-indexed) of d:
Use the bitwise OR (|) operation with a bitmask where only the c-th bit is set. This is constructed as (1<<c).
The result is d with the c-th bit set.

<br><br>
Time complexity: O(1)<br>
Space Complexity: O(1)</p>

```python
class Solution:
    def XOR(self, n, m):
        # XOR operation between n and m
        return n ^ m

    def check(self, a, b):
        # Check if the a-th bit (1-indexed) of b is set
        return 1 if (b & (1 << (a - 1))) != 0 else 0

    def setBit(self, c, d):
        # Set the c-th bit (0-indexed) of d
        return d | (1 << c)
```

<h3>5. Maximum XOR of two numbers in an array</h3>
<a href="https://leetcode.com/problems/maximum-xor-of-two-numbers-in-an-array/">Problem Link</a>
<p> 
Trie Insertion:

Each number in the array is represented in its 32-bit binary form.
We insert each bit of the number into the Trie, starting from the most significant bit (31st) to the least significant bit (0th).
Finding Maximum XOR:

For each number, traverse the Trie, and at each bit position, try to find the opposite bit (0 if the bit is 1, and 1 if the bit is 0) to maximize the XOR value.
If the opposite bit exists in the Trie, we move to that branch and record a 1 in the result for that bit position. Otherwise, we continue with the existing bit branch.
Result Computation:

The maximum XOR value is updated for each number as we compute its XOR with all previously inserted numbers in the Trie.
<br><br>
Time complexity: O(n)<br>
Space Complexity: O(n)</p>

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        
class Solution:
    def findMaximumXOR(self, nums) -> int:
        # Helper function to insert a number into the Trie
        def insert(num):
            node = root
            for i in range(31, -1, -1):  # 32-bit representation
                bit = (num >> i) & 1
                if bit not in node.children:
                    node.children[bit] = TrieNode()
                node = node.children[bit]
        
        # Helper function to find the maximum XOR with a given number
        def find_max_xor(num):
            node = root
            max_xor = 0
            for i in range(31, -1, -1):
                bit = (num >> i) & 1
                # Try to find the opposite bit for max XOR
                opposite_bit = 1 - bit
                if opposite_bit in node.children:
                    max_xor = (max_xor << 1) | 1
                    node = node.children[opposite_bit]
                else:
                    max_xor = (max_xor << 1)
                    node = node.children[bit]
            return max_xor
        
        # Initialize Trie
        root = TrieNode()
        
        # Insert all numbers into the Trie
        for num in nums:
            insert(num)
        
        # Find the maximum XOR for each number
        max_result = 0
        for num in nums:
            max_result = max(max_result, find_max_xor(num))
        
        return max_result
```

<h3>6. Maximum XOR With an Element From Array</h3>
<a href="https://leetcode.com/problems/maximum-xor-with-an-element-from-array/">Problem Link</a>
<p> 
Sorting:

Sorting nums ensures we process elements in ascending order.
Sorting queries by mi ensures that we add only the necessary elements into the Trie for each query.

Trie Operations:

insert: Inserts a number into the Trie in O(32) time.
find_max_xor: Traverses the Trie to find the number that maximizes XOR for x, also in O(32).

Query Processing:
For each query, process only relevant numbers (nums[j] <= mi), ensuring efficient handling of constraints.
<br><br>
Time complexity: O((n+q)log(n+q))<br>
Space Complexity: O(n)</p>

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.value = None  # Optional: to store value if needed (not used here)

class Solution:
    def maximizeXor(self, nums, queries):
        # Insert number into the Trie
        def insert(num):
            node = root
            for i in range(31, -1, -1):  # 32-bit integer
                bit = (num >> i) & 1
                if bit not in node.children:
                    node.children[bit] = TrieNode()
                node = node.children[bit]
        
        # Find max XOR for a given number
        def find_max_xor(x):
            if not root.children:
                return -1  # If no numbers in Trie
            node = root
            max_xor = 0
            for i in range(31, -1, -1):
                bit = (x >> i) & 1
                # Try to take the opposite bit to maximize XOR
                opposite_bit = 1 - bit
                if opposite_bit in node.children:
                    max_xor = (max_xor << 1) | 1
                    node = node.children[opposite_bit]
                else:
                    max_xor = (max_xor << 1)
                    node = node.children[bit]
            return max_xor
        
        # Sort nums and queries
        nums.sort()
        sorted_queries = sorted(enumerate(queries), key=lambda x: x[1][1])
        results = [-1] * len(queries)
        root = TrieNode()
        
        # Process queries
        idx = 0
        for q_index, (x, m) in sorted_queries:
            # Insert all nums <= m into Trie
            while idx < len(nums) and nums[idx] <= m:
                insert(nums[idx])
                idx += 1
            # Find the max XOR for the current query
            results[q_index] = find_max_xor(x)
        
        return results
```
